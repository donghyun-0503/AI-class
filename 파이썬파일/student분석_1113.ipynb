## 학생의 공부시간, 출석률, 과제점수 등의 학습데이터 분석

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../데이터파일/student.csv")
df.head()
```

## 공부 시간의 분포 알아보기

```python
plt.hist(df["StudyHours"])
plt.xlabel("Study Hours")
plt.ylabel("Frequency")
plt.title("Study Hours Distribution")
plt.show()
```

## 공부시간과 과제점수의 관계 알아보기: 산점도로

```python
plt.scatter(df["StudyHours"], df["AssignmentScore"])

plt.xlabel("Study Hours")
plt.ylabel("Assignment Score")
plt.title("Study Hours vs Assignment Score")
plt.show()
```

## 6 통과 미통과 평균 구하기

```python
result_mean = df.groupby("Result")["StudyHours"].mean()
print(result_mean)
```

```python
plt.bar(result_mean.index, result_mean.values)
plt.xlabel("Result")
plt.xticks([0, 1])
plt.ylabel("Average Study Hours")
plt.title("Average Study Hours by Result")
plt.show()
```
